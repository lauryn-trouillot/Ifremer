```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 📌 Totaux connus par organisme
totaux_transcrits = {
    "Prymnesium": 53242,
    "Dinophysis": 80589,
    "Karlodinium": 122682,
    "Lingulodinium": 94373
}

# 📁 Fichiers CSV à charger
fichiers = {
    "Lingulodinium": "/home/lauryn/Documents/Orson_results/Lingulodinium_annotations.csv",
    "Prymnesium": "/home/lauryn/Documents/Orson_results/Prymnesium_annotations.csv",
    "Dinophysis": "/home/lauryn/Documents/Orson_results/Dinophysis_annotations.csv",
    "Karlodinium": "/home/lauryn/Documents/Orson_results/Karlodinium_annotations.csv"
}
```

```python
# 🔄 Chargement et fusion
dfs = []

for organisme, chemin in fichiers.items():
    df = pd.read_csv(chemin, comment='#', sep="\t", header=None)
    df["Organisme"] = organisme
    dfs.append(df)

# Fusionner tous les DataFrames de la liste en un seul grand DataFrame
dfs = pd.concat(dfs, ignore_index=True)

dfs.columns = [
    "Sequence_ID",
    "Seed_Ortholog",
    "E_value",
    "Score",
    "eggNOG_OGs",
    "Max_Annotation_Level",
    "COG_Category",
    "Function_Description",
    "Annotated_Gene_Name",
    "GO_Terms",
    "EC_Number",
    "KEGG_KO",
    "KEGG_Pathway",
    "KEGG_Module",
    "KEGG_Reaction",
    "KEGG_rClass",
    "KEGG_BRITE",
    "KEGG_TC",
    "CAZy",
    "BiGG_Reaction",
    "Pfam_Domains",
    "Organisme"
]
```

```python
dfs = dfs.replace('-', pd.NA)
dfs
```

```python
# Dictionnaire de correspondance code -> fonction
cog_functions = {
    'J': 'Translation, ribosomal structure and biogenesis',
    'A': 'RNA processing and modification',
    'K': 'Transcription',
    'L': 'Replication, recombination and repair',
    'B': 'Chromatin structure and dynamics',
    'D': 'Cell cycle control, cell division, chromosome partitioning',
    'Y': 'Nuclear structure',
    'V': 'Defense mechanisms',
    'T': 'Signal transduction mechanisms',
    'M': 'Cell wall/membrane/envelope biogenesis',
    'N': 'Cell motility',
    'Z': 'Cytoskeleton',
    'W': 'Extracellular structures',
    'U': 'Intracellular trafficking, secretion, and vesicular transport',
    'O': 'Posttranslational modification, protein turnover, chaperones',
    'C': 'Energy production and conversion',
    'G': 'Carbohydrate transport and metabolism',
    'E': 'Amino acid transport and metabolism',
    'F': 'Nucleotide transport and metabolism',
    'H': 'Coenzyme transport and metabolism',
    'I': 'Lipid transport and metabolism',
    'P': 'Inorganic ion transport and metabolism',
    'Q': 'Secondary metabolites biosynthesis, transport and catabolism',
    'R': 'General function prediction only',
    'S': 'Function unknown'
}

# Dictionnaire de correspondance code -> grande catégorie
cog_categories = {
    'J': 'INFORMATION STORAGE AND PROCESSING',
    'A': 'INFORMATION STORAGE AND PROCESSING',
    'K': 'INFORMATION STORAGE AND PROCESSING',
    'L': 'INFORMATION STORAGE AND PROCESSING',
    'B': 'INFORMATION STORAGE AND PROCESSING',
    'D': 'CELLULAR PROCESSES AND SIGNALING',
    'Y': 'CELLULAR PROCESSES AND SIGNALING',
    'V': 'CELLULAR PROCESSES AND SIGNALING',
    'T': 'CELLULAR PROCESSES AND SIGNALING',
    'M': 'CELLULAR PROCESSES AND SIGNALING',
    'N': 'CELLULAR PROCESSES AND SIGNALING',
    'Z': 'CELLULAR PROCESSES AND SIGNALING',
    'W': 'CELLULAR PROCESSES AND SIGNALING',
    'U': 'CELLULAR PROCESSES AND SIGNALING',
    'O': 'CELLULAR PROCESSES AND SIGNALING',
    'C': 'METABOLISM',
    'G': 'METABOLISM',
    'E': 'METABOLISM',
    'F': 'METABOLISM',
    'H': 'METABOLISM',
    'I': 'METABOLISM',
    'P': 'METABOLISM',
    'Q': 'METABOLISM',
    'R': 'POORLY CHARACTERIZED',
    'S': 'POORLY CHARACTERIZED'
}

# Ajout des colonnes

# 1. Extraire la première lettre du/des code(s) COG (certaines cellules peuvent contenir plusieurs codes, ex: "K,L")
dfs['COG_code'] = dfs['COG_Category'].str[0]  # Prend la première lettre (ou NaN si vide)

# 2. Mapper la fonction COG
dfs['COG_Function'] = dfs['COG_code'].map(cog_functions)

# 3. Mapper la grande catégorie
dfs['COG_Category'] = dfs['COG_code'].map(cog_categories)

columns = list(dfs.columns)
columns.insert(columns.index('COG_Category'), columns.pop(columns.index('COG_Function')))
dfs = dfs[columns]

dfs
```

```python
annotes = dfs['Organisme'].value_counts()
# ✅ Calcul annotés vs non annotés
stat_finale = pd.DataFrame({
    "Total_transcrits": pd.Series(totaux_transcrits),
    "Annotés": annotes
})
stat_finale["Non_annotés"] = stat_finale["Total_transcrits"] - stat_finale["Annotés"]
stat_finale["% annotés"] = (stat_finale["Annotés"] / stat_finale["Total_transcrits"] * 100).round(2)

print(stat_finale)
```
